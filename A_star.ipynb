```python
import networkx as nx
```

```python
n=10
G=nx.grid_2d_graph(n,n)
```

```python
obstacles = {(3,2),(3,3),(3,4),(3,5),(4,5),(5,5), (6,5),(6,4),(1,7),(2,7),(3,7),(4,7)}
```

```python
G.remove_nodes_from(obstacles) # remove, do not penalise
```

```python
nx.set_edge_attributes(G, 1.0, "weight")
```

```python
def manhattan(u, v):
  return abs(u[0]-v[0]) + abs(u[1]-v[1])
```

```python
path = nx.astar_path(G, (0,0), (9,9),
              heuristic=manhattan, weight="weight")
```

```python
print(len(path) - 1, "moves:", path)
```
